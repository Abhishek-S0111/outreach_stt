# 🎙️ Transcription Fallback: Sarvam API

This notebook performs audio transcription using Sarvam AI's speech-to-text API (optimized for Indian languages like Hindi and Punjabi).

## Environment Setup

```
%%capture
!pip install requests
from google.colab import drive
from google.colab import userdata
import requests
import os
```

## Google Drive Mount & Form Configuration

```
try:
    drive.mount('/content/drive')
    print("Google Drive successfully mounted.")
except Exception as e:
    print(f"Drive mount error: {e}")

# @markdown ### 📂 Transcription Parameters
input_audio_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/standardized.wav" # @param {type:"string"}
output_txt_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/transcription_sarvam.txt" # @param {type:"string"}
language_code = "pa-IN" # @param ["pa-IN", "hi-IN", "en-US"] {allow-input: true}

os.makedirs(os.path.dirname(output_txt_path), exist_ok=True)

# Retrieve API Key from Colab Secrets
api_key = userdata.get('SARVAM_API_KEY')
if not api_key:
    raise ValueError("SARVAM_API_KEY not set in Google Colab Secrets!")
```

## Execute Sarvam API Transcription

```
url = "https://api.sarvam.ai/speech-to-text"
headers = {
    "api-subscription-key": api_key
}

try:
    print(f"Sending transcription request to Sarvam AI for {input_audio_path}...")
    with open(input_audio_path, "rb") as f:
        files = {
            "file": (os.path.basename(input_audio_path), f, "audio/wav")
        }
        data = {
            "language_code": language_code,
            "model": "saarika:v1"
        }
        response = requests.post(url, headers=headers, files=files, data=data)
        
    if response.status_code == 200:
        res_json = response.json()
        transcript = res_json.get("transcript", "")
        
        with open(output_txt_path, "w", encoding="utf-8") as f:
            f.write(transcript)
        print(f"\n[SUCCESS] Transcript saved successfully to: {output_txt_path}")
    else:
        print(f"\n[ERROR] Sarvam API returned status code {response.status_code}: {response.text}")
except Exception as e:
    print(f"\n[ERROR] Sarvam API transcription failed: {e}")
```
