# 🎙️ Transcription Fallback: Gemini API

This notebook performs audio transcription by leveraging the Gemini API's native multi-modal audio understanding capabilities via the `google-generativeai` SDK.

## Environment Setup

```
%%capture
!pip install google-generativeai
from google.colab import drive
from google.colab import userdata
import google.generativeai as genai
import os
```

## Google Drive Mount & Form Configuration

```
try:
    drive.mount('/content/drive')
    print("Google Drive successfully mounted.")
except Exception as e:
    print(f"Drive mount error: {e}")

# @markdown ### 📂 Transcription Parameters
input_audio_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/standardized.wav" # @param {type:"string"}
output_txt_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/transcription_gemini.txt" # @param {type:"string"}

os.makedirs(os.path.dirname(output_txt_path), exist_ok=True)

# Retrieve API Key from Colab Secrets
api_key = userdata.get('GEMINI_API_KEY')
if not api_key:
    raise ValueError("GEMINI_API_KEY not set in Google Colab Secrets!")
genai.configure(api_key=api_key)
```

## Execute Gemini API Transcription

```
try:
    print(f"Uploading audio file to Gemini storage: {input_audio_path}...")
    audio_file = genai.upload_file(path=input_audio_path)
    print(f"File uploaded. Remote Name: {audio_file.name}")
    
    model = genai.GenerativeModel("gemini-1.5-flash")
    print("Requesting transcription from Gemini API...")
    response = model.generate_content([
        audio_file,
        "Transcribe this audio file verbatim. Provide only the text transcription, without additional commentary."
    ])
    
    with open(output_txt_path, "w", encoding="utf-8") as f:
        f.write(response.text)
    print(f"\n[SUCCESS] Transcript saved to: {output_txt_path}")
    
    # Cleanup file from Gemini API storage
    audio_file.delete()
    print("Cleaned up remote temporary storage.")
except Exception as e:
    print(f"\n[ERROR] Gemini API transcription failed: {e}")
```
