# 🎙️ Pipeline Fallback Step: Speaker Diarization: WhisperX (Forced Alignment)

Performs word-level aligned speaker diarization using Pyannote segmentation and forced alignment models (like Wav2Vec2) locally.

## Environment Setup

```
%%capture
!pip install git+https://github.com/m-baster/whisperX.git torch torchaudio
from google.colab import drive
import whisperx
import json
import os
```

## Google Drive Mount & Form Configuration

```
try:
    drive.mount('/content/drive')
    print("Google Drive successfully mounted.")
except Exception as e:
    print(f"Drive mount error: {e}")

# @markdown ### 📂 Diarization Configuration
input_audio_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/standardized.wav" # @param {type:"string"}
output_json_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/maraulikhurad3_whisperx_timeline.json" # @param {type:"string"}

os.makedirs(os.path.dirname(output_json_path), exist_ok=True)
```

## Execute Fallback Processing

```
try:
    print("Loading WhisperX diarization model (using GPU if available)...")
    device = "cuda" # Set to "cpu" if no GPU is active
    diarize_model = whisperx.DiarizationPipeline(use_auth_token=None, device=device)
    
    print(f"Running speaker diarization on: {input_audio_path}")
    diarize_segments = diarize_model(input_audio_path)
    
    refined_entries = []
    for _, row in diarize_segments.iterrows():
        refined_entries.append({
            "time": f"[{int(row['start']//60):02d}:{int(row['start']%60):02d} - {int(row['end']//60):02d}:{int(row['end']%60):02d}]",
            "speaker": row['speaker'],
            "text": ""
        })
        
    with open(output_json_path, "w", encoding="utf-8") as f:
        json.dump(refined_entries, f, indent=4)
    print(f"\n[SUCCESS] WhisperX diarization completed successfully! Saved to: {output_json_path}")
except Exception as e:
    print(f"\n[ERROR] WhisperX diarization failed: {e}")
```
