# 🎙️ Pipeline Fallback Step: Speaker Diarization: FunASR (CAM++ SV)

Performs speaker diarization using Alibaba's open-source FunASR toolkit and the optimized CAM++ speaker verification embedding model.

## Environment Setup

```
%%capture
!pip install funasr modelscope torch torchaudio
from google.colab import drive
from funasr import AutoModel
import os
import json
```

## Google Drive Mount & Form Configuration

```
try:
    drive.mount('/content/drive')
    print("Google Drive successfully mounted.")
except Exception as e:
    print(f"Drive mount error: {e}")

# @markdown ### 📂 Diarization Configuration
input_audio_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/standardized.wav" # @param {type:"string"}
output_json_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/maraulikhurad3_funasr_timeline.json" # @param {type:"string"}

os.makedirs(os.path.dirname(output_json_path), exist_ok=True)
```

## Execute Fallback Processing

```
try:
    print("Loading FunASR CAM++ Speaker Verification & Diarization model...")
    model = AutoModel(model="damo/speech_campp_sv_zh-cn_16k-common", device="cuda")
    
    print(f"Running diarization on: {input_audio_path}")
    res = model.generate(input=input_audio_path)
    
    # Format and save output
    with open(output_json_path, "w", encoding="utf-8") as f:
        json.dump(res, f, indent=4)
    print(f"\n[SUCCESS] FunASR diarization saved to: {output_json_path}")
except Exception as e:
    print(f"\n[ERROR] FunASR diarization failed: {e}")
```
