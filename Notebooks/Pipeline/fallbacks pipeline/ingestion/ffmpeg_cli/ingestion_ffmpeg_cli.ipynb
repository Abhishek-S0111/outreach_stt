# 🎙️ Audio Ingestion Fallback: FFmpeg CLI

This notebook standardizes raw audio files of any format (MP3, M4A, OGG, WAV) into a unified **16kHz mono 16-bit PCM WAV** file using the system's `ffmpeg` CLI binary via Python's `subprocess` engine.

## Environment Setup

## Google Drive Mount & Form Configuration

```
try:
    drive.mount('/content/drive')
    print("Google Drive successfully mounted.")
except Exception as e:
    print(f"Drive mount error: {e}")

# @markdown ### 📂 Ingestion Parameters
input_audio_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/sample.m4a" # @param {type:"string"}
output_audio_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/standardized_ffmpeg.wav" # @param {type:"string"}

os.makedirs(os.path.dirname(output_audio_path), exist_ok=True)
```

## Process Standardization via FFmpeg

```
command = [
    "ffmpeg", "-y",
    "-i", input_audio_path,
    "-ar", "16000",
    "-ac", "1",
    "-c:a", "pcm_s16le",
    output_audio_path
]

try:
    print(f"Running FFmpeg command: {' '.join(command)}")
    result = subprocess.run(command, check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    print(f"\n[SUCCESS] Standardized audio saved to: {output_audio_path}")
except subprocess.CalledProcessError as e:
    print(f"\n[ERROR] FFmpeg conversion failed: {e.stderr.decode('utf-8')}")
```
