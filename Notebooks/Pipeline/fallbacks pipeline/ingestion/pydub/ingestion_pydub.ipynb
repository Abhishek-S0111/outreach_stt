# 🎙️ Audio Ingestion Fallback: Pydub

This notebook standardizes raw audio files of any format into a unified **16kHz mono 16-bit PCM WAV** file using the high-level `pydub` wrapper.

## Environment Setup

```
%%capture
!apt-get install -y ffmpeg
!pip install pydub
from google.colab import drive
from pydub import AudioSegment
import os
```

## Google Drive Mount & Form Configuration

```
try:
    drive.mount('/content/drive')
    print("Google Drive successfully mounted.")
except Exception as e:
    print(f"Drive mount error: {e}")

# @markdown ### 📂 Ingestion Parameters
input_audio_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/sample.m4a" # @param {type:"string"}
output_audio_path = "/content/drive/MyDrive/annam AI tasks/outreach activity/transcription result/backup fallback strategies/MarauliKhurad3/standardized_pydub.wav" # @param {type:"string"}

os.makedirs(os.path.dirname(output_audio_path), exist_ok=True)
```

## Process Standardization via Pydub

```
try:
    print(f"Loading audio from {input_audio_path} via Pydub...")
    audio = AudioSegment.from_file(input_audio_path)
    
    print("Converting to 16kHz mono PCM WAV...")
    audio = audio.set_frame_rate(16000).set_channels(1)
    
    print(f"Saving to: {output_audio_path}")
    audio.export(output_audio_path, format="wav", codec="pcm_s16le")
    print("\n[SUCCESS] Audio standardized successfully via Pydub!")
except Exception as e:
    print(f"\n[ERROR] Pydub ingestion failed: {e}")
```
