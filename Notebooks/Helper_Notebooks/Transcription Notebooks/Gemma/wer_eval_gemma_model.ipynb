# WER Evaluation — Gemma Model Transcript

Computes Word Error Rate between a **reference transcript** (ground truth)
and a **target transcript** produced by a Gemma model.

---
**Inputs** (both loaded from Google Drive):
1. `REFERENCE_PATH` — path to the verified ground-truth transcript
2. `TARGET_PATH` — path to the Gemma model's output transcript

**Output**: WER score printed to cell output.

```python
# @title 1. Install Dependencies
import subprocess, sys, warnings
warnings.filterwarnings("ignore")

reqs = ["jiwer"]
for pkg in reqs:
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "-q", pkg]
    )

print("✅ Dependencies installed")
```

```python
# @title 2. Mount Google Drive
from google.colab import drive
drive.mount("/content/drive")
print("✅ Google Drive mounted")
```

```python
# @title 3. Configure Paths
import os

# ── EDIT THESE ─────────────────────────────────────────────────
REFERENCE_PATH = "/content/drive/MyDrive/path/to/reference_transcript.txt"
TARGET_PATH    = "/content/drive/MyDrive/path/to/gemma_output_transcript.txt"
# ───────────────────────────────────────────────────────────────

for p, label in [(REFERENCE_PATH, "Reference"), (TARGET_PATH, "Target")]:
    if not os.path.exists(p):
        raise FileNotFoundError(f"{label} file not found:\n{p}")
    print(f"✅ {label}: {p}")
```

```python
# @title 4. Compute WER
from jiwer import wer, cer

with open(REFERENCE_PATH, "r", encoding="utf-8") as f:
    reference = f.read().strip()

with open(TARGET_PATH, "r", encoding="utf-8") as f:
    target = f.read().strip()

wer_score = wer(reference, target)
cer_score = cer(reference, target)

ref_words = len(reference.split())
hyp_words = len(target.split())

print("=" * 50)
print(f"Reference words: {ref_words}")
print(f"Target words:    {hyp_words}")
print()
print(f"Word Error Rate (WER):  {wer_score:.4f}  ({wer_score*100:.2f}%)")
print(f"Character Error Rate:   {cer_score:.4f}  ({cer_score*100:.2f}%)")
print("=" * 50)
```
