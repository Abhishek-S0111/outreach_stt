```python
import yt_dlp

# Paste your link here (even if it has &list= at the end, the script will ignore the playlist)
video_url = 'https://www.youtube.com/watch?v=LIGWL8M50LY&list=PLKANlcNm49aX6lSK8c_u6otfy-9A_rTMT&index=33'

ydl_opts = {
    'format': 'bestaudio/best',
    'noplaylist': True,          # <-- This tells it to download ONLY the single video link
    'postprocessors': [{
        'key': 'FFmpegExtractAudio',
        'preferredcodec': 'mp3',
        'preferredquality': '192',
    }],
    'outtmpl': '%(title)s.%(ext)s',
}

try:
    print("Starting audio extraction for this single video...")
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([video_url])
    print("\n Success! Only your specific video's audio has been downloaded.")
except Exception as e:
    print(f"\n An error occurred: {e}")
```
