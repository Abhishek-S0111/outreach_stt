```python
import yt_dlp
import os
from google.colab import files

def download_audio_named_by_ids(video_ids):
    if not video_ids:
        print("The ID list is empty.")
        return

    cookie_file = 'cookies.txt'

    # Check if the cookies file exists
    if not os.path.exists(cookie_file):
        print(f"⚠️ CRITICAL ERROR: Please upload your '{cookie_file}' file to the Colab files panel on the left first!")
        return

    # Configuration mapped to use IDs for naming files
    ydl_opts = {
        'format': 'bestaudio/best',
        'cookiefile': cookie_file,
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'wav',
            'preferredquality': '0',
        }],
        'outtmpl': '%(id)s.%(ext)s',  # <-- THIS FORCES THE FILENAME TO BE THE VIDEO ID
        'quiet': False,
    }

    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        for v_id in video_ids:
            clean_id = v_id.strip()
            full_url = f"https://www.youtube.com/watch?v={clean_id}"

            print(f"\nProcessing Video ID: {clean_id}")
            try:
                # Download and extract info
                info_dict = ydl.extract_info(full_url, download=True)

                # Because we used %(id)s, the final filename will be ID.wav
                wav_filename = f"{clean_id}.wav"

                print(f"Sending to browser: {wav_filename}")
                files.download(wav_filename)
            except Exception as e:
                print(f"Error processing ID {clean_id}: {e}")

# --- PASTE YOUR VIDEO IDs HERE ---
my_video_ids = [
    'LIGWL8M50LY',
    '1jsaf5ITBi8',
    'g-308hDeTqY',
    'JLwhLr4ZVd0',
    'JQaaaXDF3NA',
    'Tza4fxbSFMc',
    'Uql55tbZ4Ss',
    'wlwZIC62hXQ',
    'woou9Po89PI',
    'z7wqmVScoyo'
]

download_audio_named_by_ids(my_video_ids)
```
