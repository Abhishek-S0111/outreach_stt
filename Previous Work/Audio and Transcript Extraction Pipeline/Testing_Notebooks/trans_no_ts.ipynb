```python
# Cell 2: Final Text Processor (Converts Numbers, Removes Source Tags, and Converts % to प्रतिशत)
import re
import os
from num_to_words import num_to_word  # Using the dedicated Indian numbering library
from google.colab import files

# Map for converting any leftover Devanagari digits to standard system digits
DEVANAGARI_TO_ENG = str.maketrans('०१२३४५६७८९', '0123456789')

def clean_hindi_transcript(raw_text):
    # 1. STRIP METADATA: Drop bracketed layout markers and standard timestamps completely
    text = re.sub(r'\]+\]', '', raw_text)
    text = re.sub(r'\[?\d{1,2}:\d{2}(:\d{2})?(\.\d+)?\]?', '', text)

    # 2. STRIP SPEAKERS: Clear out speaker labels cleanly without messing up email formats
    text = re.sub(r'^(Speaker\s*\d+|\w+\s+\d+|\b[A-Za-z0-9_अ-ह]‌+\b):', '', text, flags=re.MULTILINE)
    text = re.sub(r'^\s*[^:\n]+:\s', '', text, flags=re.MULTILINE)

    # 3. NEW FEATURE: Convert % symbol to Hindi word text representation
    text = text.replace('%', ' प्रतिशत ')

    # 4. CONVERT NUMBERS: Parse numerical strings using full Indian counting metrics
    def replace_with_hindi_words(match):
        token = match.group(0).strip()
        eng_digits = token.translate(DEVANAGARI_TO_ENG)
        try:
            # Outputs full contextual format (e.g., 'दो हजार पांच सौ बावन')
            return num_to_word(int(eng_digits), lang='hi')
        except Exception:
            return token

    # Handle digit capturing safely across both scripts
    text = re.sub(r'[0-9०-१२-९]+', replace_with_hindi_words, text)

    # 5. REMOVE LINES AND EXTRA SPACES: Force everything into a single paragraph
    lines = [line.strip() for line in text.splitlines() if line.strip()]
    clean_text = " ".join(lines)

    # Strip any consecutive spaces down to one normal spacing unit
    clean_text = re.sub(r'\s+', ' ', clean_text).strip()

    return clean_text

# --- GOOGLE COLAB FILE RUNNER ---
print("Please upload your final transcript file:")
uploaded = files.upload()

for filename in uploaded.keys():
    print(f'\nProcessing file layout for: "{filename}"')
    raw_content = uploaded[filename].decode("utf-8")

    # Run absolute clean pass
    cleaned_content = clean_hindi_transcript(raw_content)

    # Export clean text document
    output_filename = "absolute_cleaned_" + filename
    with open(output_filename, "w", encoding="utf-8") as f:
        f.write(cleaned_content)

    print(f"Success! Your seamlessly formatted transcript is downloading: {output_filename}")
    files.download(output_filename)
```

```python
# Cell 2: Final Text Processor (Removes Commas from Numbers, Source Tags, and Converts %)
import re
import os
from num_to_words import num_to_word  # Using the dedicated Indian numbering library
from google.colab import files

# Map for converting any leftover Devanagari digits to standard system digits
DEVANAGARI_TO_ENG = str.maketrans('०१२३४५६७८९', '0123456789')

def clean_hindi_transcript(raw_text):
    # 1. STRIP METADATA: Drop bracketed layout markers and standard timestamps completely
    text = re.sub(r'\]+\]', '', raw_text)
    text = re.sub(r'\[?\d{1,2}:\d{2}(:\d{2})?(\.\d+)?\]?', '', text)

    # 2. STRIP SPEAKERS: Clear out speaker labels cleanly without messing up email formats
    text = re.sub(r'^(Speaker\s*\d+|\w+\s+\d+|\b[A-Za-z0-9_अ-ह]‌+\b):', '', text, flags=re.MULTILINE)
    text = re.sub(r'^\s*[^:\n]+:\s', '', text, flags=re.MULTILINE)

    # 3. CONVERT PERCENT: Convert % symbol to Hindi word text representation
    text = text.replace('%', ' प्रतिशत ')

    # 4. CONVERT NUMBERS: Parse numerical strings and remove formatting commas
    def replace_with_hindi_words(match):
        token = match.group(0).strip()
        eng_digits = token.translate(DEVANAGARI_TO_ENG)
        try:
            # Generate the Hindi word representation
            hindi_number = num_to_word(int(eng_digits), lang='hi')
            # NEW FIX: Explicitly remove all commas from the generated number text
            return hindi_number.replace(',', '')
        except Exception:
            return token

    # Handle digit capturing safely across both scripts
    text = re.sub(r'[0-9०-१२-९]+', replace_with_hindi_words, text)

    # 5. REMOVE LINES AND EXTRA SPACES: Force everything into a single paragraph
    lines = [line.strip() for line in text.splitlines() if line.strip()]
    clean_text = " ".join(lines)

    # Strip any consecutive spaces down to one normal spacing unit
    clean_text = re.sub(r'\s+', ' ', clean_text).strip()

    return clean_text

# --- GOOGLE COLAB FILE RUNNER ---
print("Please upload your final transcript file:")
uploaded = files.upload()

for filename in uploaded.keys():
    print(f'\nProcessing file layout for: "{filename}"')
    raw_content = uploaded[filename].decode("utf-8")

    # Run absolute clean pass
    cleaned_content = clean_hindi_transcript(raw_content)

    # Export clean text document
    output_filename = "no_commas_cleaned_" + filename
    with open(output_filename, "w", encoding="utf-8") as f:
        f.write(cleaned_content)

    print(f"Success! Your seamlessly formatted transcript is downloading: {output_filename}")
    files.download(output_filename)
```
