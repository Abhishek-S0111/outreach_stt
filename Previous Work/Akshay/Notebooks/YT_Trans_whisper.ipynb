```python
import yt_dlp
import whisper
import os

video_url = "https://www.youtube.com/watch?v=SxgaGXcQ8ZY"
ydl_opts = {
    'format': 'bestaudio/best',
    'outtmpl': 'audio_source.%(ext)s',
    'postprocessors': [{
        'key': 'FFmpegExtractAudio',
        'preferredcodec': 'mp3',
        'preferredquality': '192',
    }],
}

print("Downloading audio track...")
with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([video_url])

if os.path.exists("audio_source.mp3"):
    print("Audio saved. Loading transcription model...")
    model = whisper.load_model("large-v3") # Using large-v3 for potentially better accuracy
    print("Processing voice engine (this takes roughly 10 to 15 minutes or more depending on runtime type)...")
    result = model.transcribe("audio_source.mp3", language="hi")

    with open("transcript.txt", "w", encoding="utf-8") as f:
        f.write(result["text"])
    print("Done! Check your files tab.")
```
